**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__ (self, balance):
        self.__balance = balance

    def deposit(self,deposit_amnt):
        self.__balance += deposit_amnt
        return f"""your deposite amount is Rs:{deposit_amnt} current account balance is Rs:{self.__balance}"""

    def withdraw(self,withdraw_amnt):
        if self.__balance >= withdraw_amnt:
            self.__balance -= withdraw_amnt
            return f"your withdrawal Rs:{withdraw_amnt} successfully completed. Current balance is Rs:{self.__balance}"
        else:
            print("your balance is not enough")
```

```python
c1 = BankAccount(10000)
c1.deposit(20000)
```

```python
c1.withdraw(10000)
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Write your code here for Q4
class Phone:
    def __init__(self):
        self.price = 200000
        self._model = "s-26 ultra"  # protected variable
        self.__phone = "samsung"    # private variable

    def phone_name(self):
        return self.__phone

p1 = Phone()

print(f"price: {p1.price}")

print(f"model: {p1._model}")

try:
    print(p1.__phone)
except AttributeError as a:
    print("error:", a)

print(f"phone: {p1.phone_name()}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def __init__(self, name):
        self.name = name
    def display_info(self):
        print("This shape is:",self.name)


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius
        
    def area(self):
        cal_area = 3.1415 * (self.radius ** 2)
        print(f"{self.name} Radius: {self.radius} | Area: {cal_area:.2f}")


class Square(Shape):
    def __init__(self, side_lent):
        super().__init__("Square")
        self.side_lent = side_lent

    def area(self):
        cal_area = self.side_lent ** 2
        print(f"{self.name} Side Length: {self.side_lent} | Area: {cal_area}")



# object for Circle
c = Circle(5)
c.display_info()
c.area()          # Circle's method
print("-" * 40)
# object for Square
s = Square(4)
s.display_info()
s.area()          # Square's method
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Bill:
    def __init__(self, products_price):
        self.products_price = products_price

    def __add__(self, shopping_bag_price):
        return self.products_price + shopping_bag_price.products_price

b1 = Bill(10000)
b2 = Bill(500)

b3 = b1 + b2
print("Total price:",b3)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod
class Car:
    def start_engine(self):
        print("engine is starting")
    @abstractmethod
    def drive(self):
        pass

class toyota(Car):
    def drive(self):
        print("car ready to drive")

t1 = toyota()
t1.start_engine()
t1.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def payment_process(self, amount):
        pass
        
class CreditCardPayment(Payment):
    def payment_process(self, amount):
        print(f"your payment rupess: {amount} throght Credit Card")
    
class UPIPayment(Payment):
    def payment_process(self, amount):
        print(f"your payment rupess: {amount} throght UPI")

cc = CreditCardPayment()
cc.payment_process(5000)
print('-' * 45)
upi = UPIPayment()
upi.payment_process(1500)
```
